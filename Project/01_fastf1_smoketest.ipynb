```python
import os
import fastf1
import pandas as pd
import numpy as np

ROOT = os.getcwd()  # because the notebook is inside ResearchProject
CACHE_DIR = os.path.join(ROOT, "fastf1_cache")
os.makedirs(CACHE_DIR, exist_ok=True)

# Enable caching
fastf1.Cache.enable_cache(CACHE_DIR)

print("Project folder:", ROOT)
print("FastF1 cache:", CACHE_DIR)
```
